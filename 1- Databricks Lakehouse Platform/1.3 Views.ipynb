## Creating Stored Views
Stoké tout le temps

```sql
CREATE OR REPLACE VIEW view_apple_phones
AS SELECT *
    FROM smartphones
    WHERE brand = 'Apple';

SELECT * FROM view_apple_phones
```

# Creating Temporary Views
Stocké par session

```sql
CREATE OR REPLACE TEMP VIEW temp_view_phones_brands
AS SELECT DISTINCT brand
    FROM smartphones;

SELECT * FROM temp_view_phones_brands
```

## Creating Global Temporary Views
Stoké par cluster

```sql
CREATE OR REPLACE VIEW latest_phones
AS SELECT * FROM smartphones
    WHERE year > 2020
    ORDER BY year DESC;

SELECT * FROM latest_phones;
```
