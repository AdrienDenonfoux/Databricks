```sql
UPDATE employees 
SET salary = salary + 100
WHERE name LIKE "A%"
```

## Commande utile

_delta_log est un repertoire stockant les fichiers de gestions de la delta table

```sql
%fs ls 'dbfs:/user/hive/warehouse/employees/_delta_log'
```

## VACUUM Command

Permet de nettoyer les fichiers de gestion d'une delta table (plus de 7 jours après création)

```sql
VACUUM employees RETAIN 168 HOURS
```

# MANAGED OR EXTERNAL TABLE

```sql
CREATE TABLE managed_default
  (width INT, length INT, height INT);

INSERT INTO managed_default
VALUES (3 INT, 2 INT, 1 INT)
```

```sql
DESCRIBE EXTENDED managed_default
```

```sql
CREATE OR REPLACE TABLE external_default
  (width INT, length INT, height INT)
LOCATION 'abfss://unity-catalog-storage@dbstoragetvt2q5j2w2v26.dfs.core.windows.net/7405618176164444/external_default';

INSERT INTO external_default
VALUES (3 INT, 2 INT, 1 INT)
```

```sql
DESCRIBE EXTENDED external_default
```

```sql
CREATE SCHEMA new_default
```

```sql
DESCRIBE DATABASE EXTENDED new_default
```

```sql
USE new_default;

CREATE TABLE managed_new_default
  (width INT, length INT, height INT);
  
INSERT INTO managed_new_default
VALUES (3 INT, 2 INT, 1 INT);

-----------------------------------

CREATE TABLE external_new_default
  (width INT, length INT, height INT)
LOCATION 'abfss://unity-catalog-storage@dbstoragetvt2q5j2w2v26.dfs.core.windows.net/7405618176164444/external_new_default';
  
INSERT INTO external_new_default
VALUES (3 INT, 2 INT, 1 INT);
```

```sql
DESCRIBE EXTENDED managed_new_default
```

```sql
DESCRIBE EXTENDED external_new_default
```
