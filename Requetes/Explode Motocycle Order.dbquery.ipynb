```sql
WITH exploded_orders AS (
  SELECT
    m.motocycle_id,
    m.quantity,
    m.subtotal,
    o.order_id,
    o.customer_id,
    o.timestamp,
    o.total
  FROM demoworkspace.motocycle_bundle.orders_raw o
  LATERAL VIEW explode(o.motocycles) exploded AS m
)
SELECT
    o.*,
    m.brand,
    m.name,
    m.category,
    m.engine_displacement_cc,
    m.horsepower,
    m.price
FROM exploded_orders o
INNER JOIN demoworkspace.default.motocycles m
ON m.motocycle_id = o.motocycle_id
```
